# Student Placement Data Analysis
Exploratory Data Analysis (EDA) of student placement outcomes, academic performance, work experience, specialization, and salary.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Load the Dataset

```python
placement_data = pd.read_csv('Placement.csv')
placement_data.head()
```

```python
print('Shape:', placement_data.shape)
placement_data.info()
```

## 2. Data Cleaning and Missing Values

```python
placement_data['salary'] = placement_data['salary'].str.replace(',', '', regex=False).astype(float)
placement_data.isnull().sum()
```

```python
not_placed_missing_salary = placement_data[(placement_data['status'] == 'Not Placed') & (placement_data['salary'].isna())]
print('Not placed students with missing salary:', len(not_placed_missing_salary))
placement_data['salary'] = placement_data['salary'].fillna(0)
```

## 3. Descriptive Statistics

```python
placement_data.describe(include='all')
```

## 4. Categorical Distributions

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
columns = ['gender', 'hsc_s', 'degree_t', 'work_ex', 'specialisation', 'status']
for ax, col in zip(axes.flatten(), columns):
    sns.countplot(data=placement_data, x=col, ax=ax)
    ax.set_title(f'Distribution of {col}')
    ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

## 5. Numerical Distributions and Outliers

```python
numeric_cols = ['hsc_p', 'degree_p', 'etest_p', 'salary']
placement_data[numeric_cols].hist(figsize=(12, 8), bins=20)
plt.tight_layout()
plt.show()
```

```python
for col in ['hsc_p', 'degree_p', 'etest_p']:
    plt.figure(figsize=(7, 4))
    sns.boxplot(data=placement_data, x=col)
    plt.title(f'Boxplot of {col}')
    plt.show()
```

## 6. Placement Status vs Academic Performance

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5))
sns.boxplot(data=placement_data, x='status', y='hsc_p', ax=axes[0])
sns.boxplot(data=placement_data, x='status', y='degree_p', ax=axes[1])
sns.boxplot(data=placement_data, x='status', y='etest_p', ax=axes[2])
axes[0].set_title('HSC Percentage by Placement Status')
axes[1].set_title('Degree Percentage by Placement Status')
axes[2].set_title('Employability Test by Placement Status')
plt.tight_layout()
plt.show()
```

## 7. Placement by Degree and MBA Specialization

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=placement_data, x='degree_t', hue='status')
plt.title('Placement Status by Degree Type')
plt.show()

plt.figure(figsize=(8, 5))
sns.countplot(data=placement_data, x='specialisation', hue='status')
plt.title('Placement Status by MBA Specialization')
plt.show()
```

## 8. Gender and Placement

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=placement_data, x='gender', hue='status')
plt.title('Placement Status by Gender')
plt.show()
```

## 9. Salary Analysis

```python
placed = placement_data[placement_data['status'] == 'Placed']
plt.figure(figsize=(9, 5))
sns.histplot(data=placed, x='salary', bins=20, kde=True)
plt.title('Salary Distribution for Placed Students')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=placed, x='work_ex', y='salary')
plt.title('Salary by Work Experience')
plt.show()

placed.groupby('work_ex')['salary'].agg(['count', 'mean', 'median'])
```

## 10. Employability Test and Salary

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=placed, x='etest_p', y='salary', hue='work_ex')
plt.title('Employability Test Score vs Salary')
plt.show()
```

## Key Takeaways
- The analysis explores how academic performance, degree type, MBA specialization, work experience, gender, and employability test scores relate to placement outcomes.
- Academic performance and degree specialization show notable relationships with placement status in this dataset.
- Salary analysis is restricted to placed students when studying salary distributions and relationships.
- These findings are exploratory associations and should not be interpreted as causal effects.
