```python
#Q1: Which factors influence a student in getting summer placement? Which ones are
#more effective & which ones are less effective?

#To assess the factors influencing a student's probability of securing a summer placement and to evaluate their effectiveness, we can examine both categorical and numerical attributes from the dataset provided in the previous analysis:
#Factors with Higher Effectiveness:
#1.	Degree Specialization (degree_t and specialisation):
#•	Students specializing in Commerce & Management (Comm&Mgmt) exhibit the highest placement rates, indicating alignment with industry requirements.
#•	Moreover, within MBA specializations, candidates in Marketing & Finance (Mkt&Fin) have a higher likelihood of placement compared to those in Marketing & HR (Mkt&HR).
#2.	Academic Performance (hsc_p, degree_p, etest_p):
#•	Enhanced academic performance in higher secondary education (hsc_p), degree programs (degree_p), and employability test scores (etest_p) positively correlate with placement probabilities. Students with superior scores in these domains generally secure placements more frequently.
#Factors with Lower Effectiveness:
#1.	Higher Secondary Specialization (hsc_s):
#•	The choice of study field at the higher secondary level (Commerce, Science, Arts) exhibits a weaker correlation with placement outcomes compared to degree specialization and scores. Although Commerce students tend to have slightly better placement rates, this factor is less decisive.
#.	Gender:•	Despite a slight predominance of males in placement, gender has a lesser impact compared to academic performance and specialization. Initiatives to ensure gender parity and equitable opportunities could help alleviate this discrepancy.
#3.	Work Experience (work_ex):
#	Contrary to common assumptions, students without prior work experience are placed more frequently than those with work experience. This could stem from factors such as the nature of roles being recruited for (entry-level vs. experienced positions), relevance of prior work experience to new roles, or emphasis on academic achievements over professional experience in certain industries.
#Insight:
#•	Academic performance at various educational stages (secondary, tertiary, and professional tests) and degree specialization, particularly in Commerce & Management and Marketing & Finance, emerge as the most influential predictors of summer placement success.
#	While factors such as higher secondary specialization, gender, and work experience retain relevance, their impact on placement outcomes is comparatively lower. This analysis underscores the importance for students aiming for successful placements to prioritize academic excellence and choose degree programs and specializations that closely align with market demands.
```

```python
#Q4: Are males hired more than females
# Filter the dataset for students who are placed
placed_students = placement_data[placement_data['status'] == 'Placed']

# Count the number of placed students by gender
gender_counts = placed_students['gender'].value_counts()

# Print the counts
print(gender_counts)


# Create a bar plot for the number of placed students by gender
plt.figure(figsize=(8, 6))
gender_counts.plot(kind='bar', color=['skyblue', 'pink'])  # Assuming 'blue' for males and 'pink' for females
plt.title('Number of Placed by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Placements')
plt.xticks(rotation=0)  # Keeps the labels horizontal
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#Q7:Did people with work_ex performed better in employbility test than ones with no experience?
placement_data.groupby(['specialisation','work_ex']).etest_p.mean()
```

```python
# Group by specialization and work experience, calculate average employability test scores
etest_scores = placement_data.groupby(['specialisation', 'work_ex'])['etest_p'].mean().unstack()

# Plot
plt.figure(figsize=(10, 6))
ax = etest_scores.plot(kind='bar', color=['skyblue', 'lightgreen'])
plt.title('Average Employability Test Scores by Specialization and Work Experience')
plt.xlabel('Specialization')
plt.ylabel('Average Employability Test Score')
plt.xticks(rotation=0)
plt.legend(title='Work Experience', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
